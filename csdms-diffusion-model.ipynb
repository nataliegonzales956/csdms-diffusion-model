# A 1-D diffusion model

here we develop a 1-d nurmeical model of diffusion 

Here's the diffusion equation
$$ \frac{\partial z}{\partial t} = D\frac{\partial^2 z}{\partial x^2} $$

Here $z$ is elevation, $x$ is the spatial dimension, and $D$ is the transport rate coefficent (of Diffusivity)

the discretized version of this equation is: 
$$ z^{t+1}_x = z^t_x + {D \Delta t \over \Delta x^2} (z^t_{x+1} - 2z^t_x + z^t_{x-1}) $$

This is the FTCS scheme for soliving the diffusion equation.

start by setting two fixed model parameters, the diffusivity and the size of the model domain.

```python
D = 100 # diffusivity
Lx = 300 # model domain
```

Next, set up the model grid using a numpy array

```python
dx = 0.5 # grid spacing 
x = np.arange(start=0, stop=Lx, step=dx) # the grid
nx = len(x) # number of grid nodes
```

set inital conditions for the model. the elevation $z$ is a step function with a high value on the left, a low value on the right, and a cliff at the center of the domain



```python
z = np.zeros_like(x) #elevation
z_left= 500
z_right = 0
z[x <= Lx/2] = z_left # step
z[x > Lx/2] = z_right # step
```

make a plot of the inti

```python
plt.figure
plt.plot(x,z, "b")
plt.xlabel('x')
plt.ylabel('z')
plt.title("initial 
diffusion plot")
```

set the number of time steps in the model 
calculate a stable  time using a stability criterion


```python
nt = 5000 # number of time steps 
dt = 0.5 * dx**2 / D # time step
```

```python
for t in range(0, nt):
	z[1:-1] += D * dt / dx ** 2 * (z[:-2] - 2*z[1:-1] + z[2:])
    
```

Note that this is an array operation on th elevation $z$. the boundary condistions remia fized in each step.
